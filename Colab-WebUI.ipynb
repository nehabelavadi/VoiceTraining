```
import csv
import os

metadata = "/content/dataset_jeet/training_clips/metadata.csv"
output = "/content/dataset_jeet/jeet.list"

with open(metadata, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    with open(output, "w", encoding="utf-8") as out:
        for row in reader:
            audio = os.path.abspath(row["audio_path"])
            text = row["text"].strip()

            if os.path.exists(audio) and text:
                out.write(f"{audio}|jeet|en|{text}\n")

print("GPT-SoVITS list created!")
print("Location:", output)

with open(output, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("Entries:", len(lines))
print("\nFirst entry:")
print(lines[0])
```

```
import csv
import os

metadata = "/content/dataset_jeet/training_clips/metadata.csv"
output = "/content/dataset_jeet/jeet.list"

with open(metadata, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    with open(output, "w", encoding="utf-8") as out:
        for row in reader:
            relative_path = row["audio_path"].replace("\\", "/")

            # Convert dataset_jeet/... into /content/dataset_jeet/...
            if relative_path.startswith("dataset_jeet/"):
                audio = "/content/" + relative_path
            else:
                audio = os.path.join(
                    "/content/dataset_jeet/training_clips",
                    os.path.basename(relative_path)
                )

            text = row["text"].strip()

            if os.path.exists(audio) and text:
                out.write(f"{audio}|jeet|en|{text}\n")

with open(output, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("GPT-SoVITS list created!")
print("Entries:", len(lines))

if lines:
    print("\nFirst entry:")
    print(lines[0])
```

```
import random
import os

list_file = "/content/dataset_jeet/jeet.list"

with open(list_file, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f if line.strip()]

random.seed(42)
random.shuffle(lines)

# 90% training, 10% validation
split = int(len(lines) * 0.9)

train_lines = lines[:split]
val_lines = lines[split:]

train_file = "/content/dataset_jeet/train.list"
val_file = "/content/dataset_jeet/val.list"

with open(train_file, "w", encoding="utf-8") as f:
    f.write("\n".join(train_lines) + "\n")

with open(val_file, "w", encoding="utf-8") as f:
    f.write("\n".join(val_lines) + "\n")

print("Dataset split completed!")
print("Training clips:", len(train_lines))
print("Validation clips:", len(val_lines))
print("Total:", len(train_lines) + len(val_lines))
```
